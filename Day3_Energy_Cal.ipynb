```python
import pandas as pd
df = pd.read_excel("Appliences.xlsx")
df.info()
```

calculating daily kwh for all appliences

```python
df["daily_kwh"] = (df["Watts"]*df["Hours Per Day"])/1000
```

converting wh to kwh



```python
(df["Watts"] * df["Hours Per Day"])/1000
```

# Round the numbers
it keeps decimal to 2 digit so output look clean

```python
df["daily_kwh"] = df["daily_kwh"].round(2)
```

```python
df["daily_kwh"].round(2)
```

To find the total and biggest consumer,
Total daily energy for household

```python
total_daily = df["daily_kwh"].sum()
print("Total Daily Energy:",round(total_daily,2),"kwh")
```

for the appliences who use most of the energy

```python
df.sort_values("daily_kwh",ascending=False)
```

Top 3 consumers

```python
df.sort_values("daily_kwh",ascending=False).head(3)
```

Calculate each applience share of the total

```python
df["percent_of_total"]=(df["daily_kwh"] / total_daily * 100).round(1)
df.sort_values("percent_of_total",ascending=False)
```

the percentage should add upto 100

```python
df["percent_of_total"].sum()
```

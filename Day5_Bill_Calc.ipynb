```python
import pandas as pd
df = pd.read_excel("appliances_clean.xlsx")
df["monthly_kwh"]=(df["daily_kwh"]*30).round(2)
total_units=df["monthly_kwh"].sum()
print("Total monthly units:",round(total_units,2))
```

***Bill Function***

```python
def calculate_bill(units):
  rate1,rate2,rate3,rate4=4,5,7,9
  fixed_charge=0
  if units<=100:
    energy_charge=units*rate1
  elif units<=200:
    energy_charge=100*rate1+(units-100)*rate2
  elif units<=300:
    energy_charge=100*rate1+100*rate2+(units-200)*rate3
  else:
    energy_charge=100*rate1+100*rate2+100*rate3+(units-300)*rate4
  return energy_charge+fixed_charge
```

Estimating my household bill

```python
estimated_bill=calculate_bill(total_units)
print("Estimated monthly units:",round(total_units,2))
print("Estimated monthly bill:",round(estimated_bill,2))
```

```python
import pandas as pd
df = pd.read_excel("Appliences.xlsx")
df.head()
```

```python
df = df.rename(columns={
    "Appliences" : "appliances",
    "Watts" : "watts",
    "Hours Per Day" :"hours_per_day"
})
df.columns
```

```python
df["daily_kwh"] = (df["watts"]*df["hours_per_day"])/1000
```

```python
df["daily_kwh"].sum()
```

```python
df["monthly_kwh"] = (df["daily_kwh"]*30).round(2)
total_monthly = df["monthly_kwh"].sum()
print("Total monthly energy:",round(total_monthly,2),"kwh")
```

```python
category_map={
    "Tubelights" :"Lighting",
    "LEDs" : "Lighting",
    "Fans" : "Cooling",
    "AC" : "Cooling",
    "Refrigerator" : "Kitchen",
    "Washmachine" : "Kitchen",
    "TV" : "Entertainment",
    "Induction Stove" : "Kitchen"
}
df["category"] = df["appliances"].map(category_map)
```

```python
df[df["category"].isna()]
```

```python
df.groupby("category")["monthly_kwh"].sum()
```

```python
cat_summary=(
    df.groupby("category")["monthly_kwh"]
    .sum()
    .reset_index()
    .sort_values("monthly_kwh",ascending=False)
)
cat_summary
```

```python
summary = df.groupby("category").agg(
    appliances = ("appliances","count"),
    total_monthly_kwh = ("monthly_kwh","sum"),
    avg_watts = ("watts","mean")
).round(2)
summary.sort_values("total_monthly_kwh",ascending=False)
```

```python
summary["percent_of_total"] = (
    summary["total_monthly_kwh"]/summary["total_monthly_kwh"].sum()*100
).round(1)
summary.sort_values("percent_of_total",ascending=False)
```
